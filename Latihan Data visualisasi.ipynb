```python
# CLEANING DATA PADA TAHAP TERAKHIR WRANGLING DATA


# menghilangkan duplicate data pada tabel customer
customers_df.drop_duplicates(inplace=True)
print("jumlah duplikasi: ", customers_df.duplicated().sum())
```

```python
# karena data diatas masih mengandung banyak info penting sehingga sayang jika langsung dibuat. maka solusninya menggunakan metode imputation utk menangani missing value
customers_df.gender.value_counts()
```

```python
# menggantikan misisng values dengan prefer not to say
customers_df.fillna(value="prefer not to say", inplace=True)
customers_df.isna().sum()
```

```python
customers_df.gender.value_counts()
```

```python
# menangani inaccurate value
customers_df[customers_df.age == customers_df.age.max()]
```

```python
# LATIHAN DATA VISUALISASI
# utk menjawab ke-6 pertanyaan bisnis DICO
# 1. bagaimana performa penjualan dan revenue perusahaan dalam beberapa bulan terakhir?
# membuat sebuah df baru utk menampung info jumlah order dan total revenue tiap bulan
# mengubah frekuensi data harian menjadi bulanan
monthly_orders_df = all_df.resample(rule='ME',on='order_date').agg({
    "order_id": "nunique",
    "total_price": "sum"
})
monthly_orders_df.index = monthly_orders_df.index.strftime('%Y-%m')
monthly_orders_df = monthly_orders_df.reset_index()
monthly_orders_df.rename(columns= {
    "order_id": "order_count",
    "total_price": "revenue"
}, inplace=True)
monthly_orders_df.head()
```

```python
# RFM analysis
# recency = utk melihat kapan terakhir seorang pelanggan melakukan transaksi
# frequency= utk mengindetifikasi seberapa sering seoarang pelanggan melaukan transaksi
# monetery = utk mengidentifikasi sebrapa besar revenue yang berasal dari pelanggan tersebut
rfm_df = all_df.groupby(by="customer_id", as_index=False).agg({
    "order_date": "max", #mengambil tgl order terakhir
    "order_id": "nunique", # menghitung jumlah order
    "total_price": "sum" # menghitung jumlah revenue yg dihasilkan
})
rfm_df.columns = ["customer_id", "max_order_timestamp", "frequency", "monetary"]

# menghitung kpn terakhir pelanggan melakukan transaksi (hari)
rfm_df["max_order_timestamp"] = rfm_df["max_order_timestamp"].dt.date
recent_date = orders_df["order_date"].dt.date.max()
rfm_df["recency"] = rfm_df["max_order_timestamp"].apply(lambda x: (recent_date -x).days)
rfm_df.drop("max_order_timestamp", axis=1, inplace=True)
rfm_df.head()
```

```python
# mengidentifikasi best customer berdasarkan parameter frequency, monetary, recency
fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(30,6))
colors = ["#72BCD4","#72BCD4","#72BCD4","#72BCD4","#72BCD4"]
sns.barplot(y="recency", x="customer_id", data=rfm_df.sort_values(by="recency", ascending=True).head(5), palette=colors, ax=ax[0])
ax[0].set_ylabel(None)
ax[0].set_xlabel(None)
ax[0].set_title("By Recency (days)", loc="center", fontsize=18)
ax[0].tick_params(axis='x', labelsize=15)

sns.barplot(y="frequency", x="customer_id", data=rfm_df.sort_values(by="frequency", ascending=False).head(5), palette=colors, ax=ax[1])
ax[1].set_ylabel(None)
ax[1].set_xlabel(None)
ax[1].set_title("By Frequency", loc="center", fontsize=18)
ax[1].tick_params(axis='x', labelsize=15)

sns.barplot(y="monetary", x="customer_id", data=rfm_df.sort_values(by="monetary", ascending=False).head(5), palette=colors, ax=ax[2])
ax[2].set_ylabel(None)
ax[2].set_xlabel(None)
ax[2].set_title("By Monetary", loc="center", fontsize=18)
ax[2].tick_params(axis='x', labelsize=15)

plt.suptitle("Best Customer Based on RFM Parameters (customer_id)", fontsize=20)
plt.show()
```
