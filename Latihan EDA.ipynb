```python
# CLEANING DATA PADA TAHAP TERAKHIR WRANGLING DATA


# menghilangkan duplicate data pada tabel customer
customers_df.drop_duplicates(inplace=True)
print("jumlah duplikasi: ", customers_df.duplicated().sum())
```

```python
# menangani missing value pada tabel customer
customers_df[customers_df.gender.isna()]  # cuman nampilin data yg missing value
```

```python
# karena data diatas masih mengandung banyak info penting sehingga sayang jika langsung dibuat. maka solusninya menggunakan metode imputation utk menangani missing value
customers_df.gender.value_counts()
```

```python
# menggantikan misisng values dengan prefer not to say
customers_df.fillna(value="prefer not to say", inplace=True)
customers_df.isna().sum()
```

```python
customers_df.gender.value_counts()
```

```python
# memperbaiki kesalahan karena typo
customers_df['gender'] = customers_df['gender'].replace('prefer not to say', 'Prefer not to say')
customers_df.gender.value_counts()
```

```python
# identifikasi lagi missing value nya
customers_df.isna().sum()
```

```python
# menangani inaccurate value
customers_df[customers_df.age == customers_df.age.max()]
```

```python
# karena age dari kolom 967 adalah 700 (maybe kesalahan manusia kelebihan 0) maka gunakan replace method
# customers_df.age.replace(customers_df.age.max(), 70, inplace=True)
# cara yg lebih aman dari code diatas karena operasi tersebut dianggap mengubah copy dari data, bukan DataFrame aslinya.
customers_df['age'] = customers_df['age'].replace(customers_df['age'].max(), 70)
```

```python
customers_df[customers_df.age == customers_df.age.max()]
```

```python
customers_df['age'] = customers_df['age'].replace(customers_df['age'].max(), 50)
```

```python
# memastikan kembali tidak inacurate pada customer
customers_df.describe()
```

```python
# membersihkan data order dengan mengubah tipe data order_Date dan delievery_date
datetime_columns = ["order_date", "delivery_date"]
for column in datetime_columns:
    orders_df[column] = pd.to_datetime(orders_df[column])
```

```python
orders_df.info()
```

```python
# membersihkan data product (data duplikat)
product_df.drop_duplicates(inplace=True)
```

```python
print("jumlah duplikasi:", product_df.duplicated().sum())
```

```python
# membersihkan data sales yg missing value
sales_df[sales_df.total_price.isna()]
```

```python
sales_df["total_price"] = sales_df["price_per_unit"] * sales_df["quantity"]
sales_df.isna().sum()
```

```python
#  LATIHAN EDA (EXPLARATORY DATA ANALYSIS)

# melihat rangkuman parameter statistik dari data customers_df
customers_df.describe(include="all")
```

```python
# melihat demografi pelanggan berdasarkan jenis kelamin dari tabel customer
customers_df.groupby(by="gender").agg({
    "customer_id": "nunique",
    "age": ["max","min","mean","std"]
})
```

```python
#  melihat persebaran jumlah pelanggan berdasarkan kota dan negara bagian
customers_df.groupby(by="city").customer_id.nunique().sort_values(ascending=False)
```

```python
#  melihat persebaran jumlah pelanggan berdasarkan negara bagian
customers_df.groupby(by="state").customer_id.nunique().sort_values(ascending=False)
```

```python
# esklorasi data orders
# membuat kolom baru utk memuat waktu pengiriman tiap order
delivery_time = orders_df["delivery_date"] - orders_df["order_date"]
delivery_time = delivery_time.apply(lambda x: x.total_seconds())
orders_df["delivery_time"] = round(delivery_time/86400)

orders_df.head()
```

```python
# melihat rangkuman parameter statistik order
orders_df.describe(include="all")
```

```python
# eksplorasi data orders_df dan customers_df
# utk mengindentifikasi pelanggan yg belom pernah melalkukan order dengan menambahkan kolom baru "status" 
# active utk yg pernah setidaknya sekali
# non active yg tidak pernah sama seklai
customer_id_in_ordes_df = orders_df.customer_id.tolist()
customers_df["status"] = customers_df["customer_id"].apply(lambda x: "Active" if x in customer_id_in_ordes_df else "Non Active")
customers_df.sample(5)
```

```python
# utk memperoleh info terkait jumlah pelanggan yg berstatus active dan non-active
customers_df.groupby(by="status").customer_id.count()
```

```python
# diketahui bahwa 62% customer tidak melakukan transaksi sebelumnya
# utk memperoleh info lebih banyak terkait kedua data [order dan customer]
orders_customers_df = pd.merge(
    left = orders_df,
    right = customers_df,
    how = "left",
    left_on = "customer_id",
    right_on = "customer_id"
)
orders_customers_df.head()
```

```python
# utk memperoleh info terkait jumlah order berdasarkan kota 
orders_customers_df.groupby(by="city").order_id.nunique().sort_values(ascending=False).reset_index().head(10)
```

```python
# jumlah order berdasarkan state
orders_customers_df.groupby(by="state").order_id.nunique().sort_values(ascending=False)
```

```python
# jumlah order berdasarkan gender
orders_customers_df.groupby(by="gender").order_id.nunique().sort_values(ascending=False)
```

```python
# jumlah order bedasarkan kelompok usia
# mengelompokan pelanggan dlm 3 kelompok yaitu youth, adult,senior dengan menambahkan kolom baru
orders_customers_df["age_group"] = orders_customers_df.age.apply(lambda x:"Youth" if x <= 24 else ("Seniors" if x > 64 else "Adults"))
orders_customers_df.groupby(by="age_group").order_id.nunique().sort_values(ascending=False)
```

```python
# eksplorasi data product 
product_df.describe(include="all")
```

```python
# eksplorasi data sales

sales_df.describe(include="all")
```

```python
# melihat produk termahal
product_df.sort_values(by="price", ascending=False)
```

```python
# mencari info terkait produk dnegan tipe dan nama produknya
product_df.groupby(by="product_type").agg({
    "product_id":"nunique",
    "quantity": "sum",
    "price":["min","max"]
})
```

```python
# mencari info terkait produk dnegan nama produknya
product_df.groupby(by="product_name").agg({
    "product_id": "nunique",
    "quantity":"sum",
    "price":["min","max"]
})
```

```python
# melihat produk yg paling laris
sales_product_df = pd.merge(
    left = sales_df,
    right = product_df,
    how = "left",
    left_on = "product_id",
    right_on = "product_id"
)
sales_product_df.head()
```

```python
# melihat info penjualan produk berdasarkan tipenya
sales_product_df.groupby(by="product_type").agg({
    "sales_id": "nunique",
    "quantity_x": "sum",
    "total_price": "sum"
})
```

```python
# utk melihat info penjualan berdasarkan nama produk
sales_product_df.groupby(by="product_name").agg({
    "sales_id": "nunique",
    "quantity_x": "sum",
    "total_price": "sum"
}).sort_values(by="total_price",ascending=False)
```

```python
# eksplorasi data all_df
# utk melihat pola pembelian berdasarkan demografi pelanggan
# membuat dataframe baru utk menampung semua info dri ke-4 tabel
all_df = pd.merge(
    left=sales_product_df,
    right=orders_customers_df,
    how= "left",
    left_on="order_id",
    right_on= "order_id"
)
all_df.head()
```

```python
# utk melihat preferensi pembelian berdasarkan satu pelanggan dan tipe produk
all_df.groupby(by=["state", "product_type"]).agg({
    "quantity_x": "sum",
    "total_price": "sum"
})
```

```python
# utk mengetahui selera tipe produk pelanggan berdasarkan genre 
all_df.groupby(by=["gender","product_type"]).agg({
    "quantity_x": "sum",
    "total_price": "sum"
})
```

```python
# utk mengetahui selera tipe produk pelanggan berdasarkan kelompok usia
all_df.groupby(by=["age_group","product_type"]).agg({
    "quantity_x": "sum",
    "total_price": "sum"
})
```
